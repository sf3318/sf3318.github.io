# Bouncy Banana: New York City Unemployment Dashboard

Our project initially set out to find the relationship between different variables and unemployment. This being our first project, we did not know what we were getting ourselves into. While our team had ideas of what visuals to display, I struggled to find an answer to the "So what" about our project.

![Initial Project Designs](images/image1.jpeg)

## Evolution

At the start of the project, we had three datasets in mind:

- City of New York Job Postings (NYC Open Data)
- United States Department of Labor Unemployment Claims (FRED)
- City of New York Motor Vehicle Collisions – Crashes (NYC OpenData)

Coming from a finance background, I had always expected Excel sheets to have useful numbers that could be analyzed and turned into meaningful information. However, when I investigated City of New York Motor Vehicle Collisions, there were no numbers that I could analyze that would produce something meaningful with our dashboard.

Furthermore, when looking into City of New York Job Postings, we learned that the sheet was updated regularly, and when a job was filled, they would remove it from that dataset. This meant that while the dataset was in real time, we could not get the historical information on jobs — it was impossible to calculate yearly numbers that would have been useful with our dashboard. Although there was an opportunity to query and create our own dataset of historical job postings, it did not make sense given our three-month period. Ultimately, we decided to drop this dataset as well.

This meant that out of our three initial datasets, we only had one that was useable. Sophia and I went back to the drawing board to figure out what to do. Eventually we decided that she would focus on health data and its relation to unemployment, and I would focus on unemployment and its relation to evictions. This led me to go on a deep dive into unemployment for New York City as a whole.

One bold assumption I had was that everything was readily available on the internet. Turns out, there is not a lot of information! I had originally used the U.S. Bureau of Labor Statistics unemployment rate for New York State to create an analysis of our variables.

![BLS Unemployment Rate vs. Claims](images/image2.png)

While this certainly did work when it came to unemployment and claims, with the above picture showing a rather strong correlation, we ran into issues on our main dashboard, where we were trying to match the actual numbers as close as possible to New York City's actual unemployment rate.

## The Five-Borough Approach

What I had to do instead was go through each individual borough's unemployment data, to create an aggregate sum to be used in our analysis.

![Actual Borough Unemployment](images/image3.png)

![Our Projections](images/image4.png)

By using the eviction share percentage and multiplying it by the total unemployed calculated using the five-borough information, we were able to get relatively close to the actual numbers minus the Bronx and Manhattan.

When taking a deeper look into these two boroughs we came to the conclusion that because the Bronx is typically associated with lower class individuals, and Manhattan is associated with higher class individuals, there may be other socioeconomic factors that may affect people getting evicted other than unemployment.

## Learnings

My biggest takeaway from the project is that we can always find a "so what" in data. I remember constantly pestering Sneha about finding meaningful information in a world where information is so readily available, but throughout the project I found holes in the vast internet that our project could fill.

Some smaller takeaways were just learning how to work with people as a whole. I have always been an advocate of doing things myself, but this project was simply impossible to do alone. Understanding the intricacies of Github (and merge conflicts!) really made me understand and appreciate collaboration as a whole!

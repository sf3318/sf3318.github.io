```python
import pandas as pd
```

### Creating Dataframe

```python
df = pd.read_csv("lol_ranked.csv")
```

### Using Pandas to calculate Mean, Median, and Mode

```python
# Blue Side Total Gold Mean,Median,Mode
blue_mean = df["blueTotalGold"].mean()
blue_median = df["blueTotalGold"].median()
blue_mode = df["blueTotalGold"].mode()
```

```python
print(f'Mean using pandas: {blue_mean:.2f}')
print(f'Median using pandas: {blue_median}')
print(f'Mode using pandas: {blue_mode[0]}, {blue_mode[1]}, {blue_mode[2]}')
```

### Calculating it the Hard Way :(

```python
# The Hard Way
data = [] # Store Data
blue_gold = [] # Store row we are looking at
position = 0
with open("lol_ranked.csv",'r') as file:
    for l in file:
        row = l.strip()
        v = row.split(",")
        data.append(v)
    while data[0][position] != 'blueTotalGold': # Get Row Position
        position += 1
    for y in data:
        blue_gold.append(y[position])
    blue_gold = blue_gold[1:]

#Solve for Mean
blue_total = 0
for x in blue_gold:
    blue_total += int(x)
hard_mean = blue_total / len(blue_gold)

#Solve for Median
blue_gold_sorted = sorted(blue_gold)
half = len(blue_gold) // 2 #Takes Middle of list
hard_median = blue_gold_sorted[half]
   

#Solve for Mode
hard_mode = [] # Store mode, if there are multiple, will store multiple values
freq = {} # Dictonary to store frequency
for v in blue_gold:
    if v in freq:
        freq[v] += 1
    else:
        freq[v] = 1
highest_freq = max(freq.values())
for x,y in freq.items():
    if y == highest_freq:
        hard_mode.append(x)
    
    
```

```python
print(f'Mean the hard way is: {hard_mean:.2f}')
print(f'Median the hard way is: {hard_median}')
print(f'Mode the hard way is: {hard_mode[0]}, {hard_mode[1]}, {hard_mode[2]}')
```
